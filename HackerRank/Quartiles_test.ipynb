```python
import sys, io
import pandas as pd
import numpy as np
```

```python
test_input = """9
3 7 8 5 12 14 21 13 18
"""

test_input
```

```python
sys.stdin = io.StringIO(test_input)
input = lambda prompt="": sys.stdin.readline().rstrip("\n")

n = int(input())
s = pd.to_numeric(pd.Series(input().split()[:n]))

x = s.sort_values().to_numpy()

q2 = pd.Series(x).median()

lower = x[:len(x)//2]
upper = x[len(x)//2 + 1:]

q1 = pd.Series(lower).median()
q3 = pd.Series(upper).median()

print(q1)
print(q2)
print(q3)
```
