```python
if len(arr) % 2 != 0:
    lower = arr[ : (len(arr)//2)]
    upper = arr[(len(arr)//2 + 1) : ]
else:
    lower = arr[ : (len(arr)//2)]
    upper = arr[(len(arr)//2) : ]

print(lower)
print(upper)
```

```python
def median(x):
    if len(x) % 2 != 0: # odd
        return x[len(x)//2]
    else: # even
        return (x[len(x)//2 - 1] + x[len(x)//2])/2
```
